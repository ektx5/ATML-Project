# 01. Exploratory Data Analysis: ASVspoof 2019 LA
This notebook analyzes class balance, attack distribution (A01-A19), and audio properties.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

protocol_path = '../datasets/ASVspoof2019_LA/ASVspoof2019_LA_cm_protocols/ASVspoof2019.LA.cm.train.trn.txt'
if os.path.exists(protocol_path):
    df = pd.read_csv(protocol_path, sep=' ', header=None, names=['speaker_id', 'utterance_id', 'env', 'attack_id', 'label'])
    print('Class counts:\n', df['label'].value_counts())
    print('Attack counts:\n', df['attack_id'].value_counts())
else:
    print('Protocol file not found at:', protocol_path)
```
