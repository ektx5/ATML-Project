# 02. Exploratory Data Analysis: VoxCeleb1
Analyzes speaker distribution, utterance counts, and verification pairs in veri_test.txt.

```python
import os
import pandas as pd

veri_test = '../datasets/VoxCeleb1/veri_test.txt'
if os.path.exists(veri_test):
    df_pairs = pd.read_csv(veri_test, sep=' ', header=None, names=['label', 'path1', 'path2'])
    print('Pairs count:', len(df_pairs))
    print('Label distribution:\n', df_pairs['label'].value_counts())
else:
    print('veri_test.txt not found at:', veri_test)
```
